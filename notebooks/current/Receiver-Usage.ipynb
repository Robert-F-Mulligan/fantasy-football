# Receiver usage by team stint

Offline migration of the target-share/air-yard-share analysis in `NFL-Weekly-Vizzes.ipynb`. The bundled plays are **synthetic** and include a transferred player and an unidentified receiver. Install `pip install -e ".[viz]"` in the kernel environment. This is descriptive research, not a projection adjustment.

```python
from pathlib import Path
import pandas as pd
from fantasyfootball.research.usage import ReceiverUsage
from fantasyfootball.reports.visualization import PlotlyVisualizer

root = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "pyproject.toml").exists() and (p / "examples").is_dir())
source = root / "examples" / "receiver_usage.csv"
plays = pd.read_csv(source, dtype={"receiver_player_id": str})
season, through_week = 2026, 2
usage, audit = ReceiverUsage().summarize(plays, season=season, through_week=through_week)
audit
```

## Denominators and coverage

Targets count only pass plays with an identified receiver; unidentified pass plays are reported separately and may include sacks or throwaways. Two-point attempts, nullified plays and postseason plays are excluded. Team stints remain separate after trades. Air yards are signed: shares can fall outside 0-100%. Incomplete or nonpositive team air-yard totals yield unknown shares, not zeros. All values use the supplied snapshot through the selected week; missing games are not inferred.

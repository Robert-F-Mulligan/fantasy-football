# Neutral-situation team tendencies

Offline migration of the neutral passing-rate analysis from `NFL-Weekly-Vizzes.ipynb`. The fixture is **synthetic**. Install `pip install -e ".[viz]"` in the selected kernel environment.

```python
from pathlib import Path
import pandas as pd
from fantasyfootball.research.tendencies import NeutralTendencies
from fantasyfootball.reports.visualization import PlotlyVisualizer

root = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "pyproject.toml").exists() and (p / "examples").is_dir())
plays = pd.read_csv(root / "examples" / "neutral_tendencies.csv")
tendencies = NeutralTendencies().summarize(plays, season=2026, through_week=1)
tendencies
```

## Definition and interpretation

Neutral means first or second down, more than 120 seconds remaining in the half, and offense win probability between 20% and 80%, inclusive. This is a declared descriptive filter, not a statistically selected threshold.

Dropbacks include sacks and scrambles. Designed rushes exclude dropbacks. Nullified plays, two-point attempts, kneels, spikes and postseason plays are excluded. Missing down, clock or win probability is counted as incomplete context and excluded from the neutral sample. The rate is play-weighted; it is not the unweighted mean of game rates. Teams with zero qualifying plays have an unknown rate and remain in the table.

These tendencies are historical and use the provider win-probability model. They do not predict next-week volume or establish an edge. Hover to see sample counts; the chart always spans 0-100%.

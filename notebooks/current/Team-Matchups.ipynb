# Team matchup research

Offline replacement for the play-by-play part of the legacy team matchup notebook. The fixture is **synthetic**. Install the package with `pip install -e ".[viz]"` in the kernel environment. All calculations use the same snapshot and explicit season/week cutoff.

```python
from pathlib import Path
import pandas as pd
from fantasyfootball.research.matchups import TeamMatchups
from fantasyfootball.reports.visualization import PlotlyVisualizer

root = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "pyproject.toml").exists() and (p / "examples").is_dir())
plays = pd.read_csv(root / "examples" / "team_matchups.csv")
service = TeamMatchups()
summary = service.summarize(plays, season=2026, through_week=1)
comparison = service.compare(summary, offense="A", defense="B")
comparison
```

## Definitions

Dropbacks include sacks and scrambles. Designed rushes exclude dropbacks, kneels and spikes. Two-point attempts, nullified plays and postseason plays are excluded. EPA is measured from the offense perspective: higher offensive EPA and higher EPA allowed by a defense favor the offense. For sacks, lower conceded and higher made favor their respective units.

Each chart has its own units. Hover shows observed and eligible play counts. Missing EPA/sack values are excluded from the metric denominator and remain visible in coverage counts. These are descriptive rates, not opponent-adjusted estimates, fantasy points allowed, or a blended game forecast. The demo samples are deliberately tiny.

```python
charts = PlotlyVisualizer()
charts.team_matchup(comparison, "Dropback EPA")
```
